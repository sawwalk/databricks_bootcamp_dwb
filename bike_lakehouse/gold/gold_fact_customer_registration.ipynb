```python
%sql
USE CATALOG databricks_bootcamp_dwb;
USE SCHEMA silver;
```

```python
%sql
SELECT * FROM crm_customers
LIMIT 5;
```

%md
# Perform Business Transformations and Modeling

```python
# Select required columns from crm_customers (only columns with = True)
df = spark.sql("""
    SELECT 
        customer_id,
        date_created
    FROM crm_customers
""")
```

%md
# Write Gold Table

```python
# Write to gold layer
df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("databricks_bootcamp_dwb.gold.fact_customer_registration")
```
