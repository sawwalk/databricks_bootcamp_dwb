%md
# Perform Business Transformations and Modeling

```python
# Select required columns from crm_products
df = spark.sql("""
    SELECT 
        product_id,
        product_key,
        product_cost,
        start_date,
        end_date_test
    FROM crm_products
""")
```

%md
# Write Gold Table

```python
# Write to gold layer
df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("databricks_bootcamp_dwb.gold.fact_product_releases")
```
