```python
%sql
USE CATALOG databricks_bootcamp_dwb;
USE SCHEMA silver;
```

```python
%sql
SELECT * FROM crm_sales
LIMIT 5;
```

# Perform Business Transformations and Modeling

```python
# Select required columns from crm_sales
df = spark.sql("""
    SELECT 
        order_number,
        product_key,
        customer_id,
        order_date,
        ship_date,
        due_date,
        sales_amount,
        quantity,
        unit_price
    FROM crm_sales
""")
```

# Write Gold Table

```python
# Write to gold layer
df.write.format("delta").mode("overwrite").saveAsTable("databricks_bootcamp_dwb.gold.fact_sales")
```

```python
%sql
SELECT *
FROM databricks_bootcamp_dwb.gold.fact_sales
LIMIT 100;
```
