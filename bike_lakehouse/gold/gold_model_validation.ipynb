## 1. Verify join types between tables 
1-to-1, 1-to-many, many-to-many

```sql
%sql --name test1_customers_registration
-- Test relationship: dim_customers.customer_id -> fact_customer_registration.customer_id

WITH dim_stats AS (
  SELECT 
    'dim_customers' AS table_name,
    COUNT(*) AS total_rows,
    COUNT(DISTINCT customer_id) AS distinct_keys,
    COUNT(*) - COUNT(DISTINCT customer_id) AS duplicate_keys
  FROM databricks_bootcamp_dwb.gold.dim_customers
),
fact_stats AS (
  SELECT 
    'fact_customer_registration' AS table_name,
    COUNT(*) AS total_rows,
    COUNT(DISTINCT customer_id) AS distinct_keys,
    COUNT(*) - COUNT(DISTINCT customer_id) AS duplicate_keys
  FROM databricks_bootcamp_dwb.gold.fact_customer_registration
)
SELECT 
  'dim_customers.customer_id -> fact_customer_registration.customer_id' AS relationship,
  d.total_rows AS dim_total_rows,
  d.distinct_keys AS dim_distinct_keys,
  d.duplicate_keys AS dim_duplicates,
  f.total_rows AS fact_total_rows,
  f.distinct_keys AS fact_distinct_keys,
  f.duplicate_keys AS fact_duplicates,
  CASE 
    WHEN d.duplicate_keys = 0 AND f.duplicate_keys = 0 THEN '1-to-1'
    WHEN d.duplicate_keys = 0 AND f.duplicate_keys > 0 THEN '1-to-many'
    WHEN d.duplicate_keys > 0 AND f.duplicate_keys = 0 THEN 'many-to-1'
    ELSE 'many-to-many'
  END AS join_type
FROM dim_stats d, fact_stats f
```

## 2. Verify the preservation of key metrics

**Metrics to verify:**

1. **Total sales amount**
   - Silver: `SUM(sales_amount)` FROM `crm_sales`
   - Gold: `SUM(sales_amount)` FROM `fact_sales`

2. **Average cost of products to manufacture**
   - Silver: `AVG(product_cost)` FROM `crm_products`
   - Gold: `AVG(product_cost)` FROM `fact_product_releases`

3. **Total customers who made a purchase**
   - Silver: `COUNT(DISTINCT customer_id)` FROM `crm_sales`
   - Gold: `COUNT(DISTINCT customer_id)` FROM `fact_sales`

4. **Minimum and maximum customer sign-ups by month**
   - Silver: Extract `MONTH(date_created)`, count customers per month, then `MIN(signup_count)` and `MAX(signup_count)` FROM `crm_customers`
   - Gold: Extract `MONTH(date_created)`, count customers per month, then `MIN(signup_count)` and `MAX(signup_count)` FROM `fact_customer_registration`

```sql
%sql --name silver_metrics
-- Calculate key metrics from SILVER layer

WITH sales_metric AS (
  SELECT 
    SUM(sales_amount) AS total_sales_amount
  FROM databricks_bootcamp_dwb.silver.crm_sales
),
product_cost_metric AS (
  SELECT 
    AVG(product_cost) AS avg_product_cost
  FROM databricks_bootcamp_dwb.silver.crm_products
),
total_customers_metric AS (
  SELECT 
    COUNT(DISTINCT customer_id) AS total_customers
  FROM databricks_bootcamp_dwb.silver.crm_sales
),
customers_by_country AS (
  SELECT 
    l.country,
    COUNT(DISTINCT l.customer_id) AS customer_count
  FROM databricks_bootcamp_dwb.silver.erp_customer_location l
  GROUP BY l.country
),
signups_by_month AS (
  SELECT 
    MONTH(date_created) AS signup_month,
    COUNT(*) AS signup_count
  FROM databricks_bootcamp_dwb.silver.crm_customers
  GROUP BY MONTH(date_created)
),
signup_extremes AS (
  SELECT 
    MIN(signup_count) AS min_signups,
    MAX(signup_count) AS max_signups
  FROM signups_by_month
)
SELECT 
  'SILVER' AS layer,
  s.total_sales_amount,
  p.avg_product_cost,
  tc.total_customers,
  se.min_signups AS min_monthly_signups,
  se.max_signups AS max_monthly_signups
FROM sales_metric s, product_cost_metric p, total_customers_metric tc, signup_extremes se
```

```sql
%sql --name gold_metrics
-- Calculate key metrics from GOLD layer

WITH sales_metric AS (
  SELECT 
    SUM(sales_amount) AS total_sales_amount
  FROM databricks_bootcamp_dwb.gold.fact_sales
),
product_cost_metric AS (
  SELECT 
    AVG(product_cost) AS avg_product_cost
  FROM databricks_bootcamp_dwb.gold.fact_product_releases
),
total_customers_metric AS (
  SELECT 
    COUNT(DISTINCT customer_id) AS total_customers
  FROM databricks_bootcamp_dwb.gold.fact_sales
),
signups_by_month AS (
  SELECT 
    MONTH(date_created) AS signup_month,
    COUNT(*) AS signup_count
  FROM databricks_bootcamp_dwb.gold.fact_customer_registration
  GROUP BY MONTH(date_created)
),
signup_extremes AS (
  SELECT 
    MIN(signup_count) AS min_signups,
    MAX(signup_count) AS max_signups
  FROM signups_by_month
)
SELECT 
  'GOLD' AS layer,
  s.total_sales_amount,
  p.avg_product_cost,
  tc.total_customers,
  se.min_signups AS min_monthly_signups,
  se.max_signups AS max_monthly_signups
FROM sales_metric s, product_cost_metric p, total_customers_metric tc, signup_extremes se
```

```sql
-- Compare silver vs gold metrics side-by-side
SELECT * FROM silver_metrics
UNION ALL
SELECT * FROM gold_metrics
```
